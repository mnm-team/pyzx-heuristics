```python
import sys; sys.path.append('../..')
import random
import pyzx as zx
from pyzx.heuristics.extraction.standardextract import HybridMappingExtractor
from pyzx.heuristics.extraction.mappers import create_na_mapper
from pyzx.heuristics.extraction.extractionutils import convert_to_qiskit, convert_mapped_circuit
import math
from fractions import Fraction
```

```python
random.seed(11)
g = zx.generate.cliffordT(qubits=4, depth=20, p_t=0.3, p_cnot=0.5)
c = zx.Circuit.from_graph(g)
c = zx.optimize.basic_optimization(c.split_phase_gates()).split_phase_gates()
print(c.stats())

zx.draw(c.to_graph())
g = c.to_graph()
zx.simplify.full_reduce(g)
zx.draw(g, scale=40, labels=True)
```

```python
mapper = create_na_mapper('rubidium.json', len(g.inputs()))
extractor = HybridMappingExtractor(g, mapper, edge_bias=0)
routed_circuit = extractor.extract(num_it=1)
```

```python
zx.draw(routed_circuit)
zx.compare_tensors(c,routed_circuit)
```

```python
qc_mapped = convert_mapped_circuit(extractor.mapper.get_mapped_qc(), len(g.inputs()), list(range(0,len(g.inputs()))))
zx.draw(qc_mapped)
zx.compare_tensors(c,qc_mapped)
```
