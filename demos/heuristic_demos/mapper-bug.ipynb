```python
def create_mapper():
    architecture_mapper = NeutralAtomHybridArchitecture('rubidium.json')
    mapper = HybridSynthesisMapper(arch=architecture_mapper)
    mapper.init_mapping(6)
    return mapper
```

```python
mapper2 = create_mapper()
```
