```python
from IPython.display import display, HTML
display(HTML("<style>.container { width:100% !important; }</style>"))
import sys; sys.path.append('../..')
import random
import pyzx as zx
from pyzx.heuristics.extraction.standardextract import HybridMappingExtractor
from pyzx.heuristics.extraction.mappers import create_na_mapper
```

```python
random.seed(36)
g = zx.generate.cliffordT(qubits=6, depth=300, p_t=0.3, p_cnot=0.5)
c = zx.Circuit.from_graph(g)
c = zx.optimize.basic_optimization(c.split_phase_gates()).split_phase_gates()
print(c.stats())

zx.draw(c.to_graph())
g = c.to_graph()
zx.simplify.full_reduce(g)
zx.draw(g, scale=40, labels=True)
```

```python
mapper = create_na_mapper('rubidium.json', len(g.inputs()))
extractor = HybridMappingExtractor(g, mapper)
routed_circuit = extractor.extract()
```

```python
zx.compare_tensors(routed_circuit,c)
```
