```python
from IPython.display import display, HTML
display(HTML("<style>.container { width:100% !important; }</style>"))
import sys; sys.path.append('../..')
import random
import pyzx as zx
from pyzx.heuristics.extraction.extraction import extract_with_mapper, create_na_mapper
```

```python
random.seed(11)
g = zx.generate.cliffordT(qubits=6, depth=80, p_t=0.3, p_cnot=0.5)
# c = zx.generate.phase_poly(n_qubits=16, n_phase_layers=20, cnots_per_layer=10)
c = zx.Circuit.from_graph(g)
c = zx.optimize.basic_optimization(c.split_phase_gates()).split_phase_gates()
print(c.stats())

zx.draw(c.to_graph())
```

```python
g = c.to_graph()
zx.simplify.full_reduce(g)
zx.draw(g)
```

```python
g_ex = g.copy()
mapper = create_na_mapper('rubidium.json',g_ex.num_inputs())
circuit = extract_with_mapper(g_ex, mapper)
```
